```python
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
df = pd.DataFrame({
    'num1': [1.0, 2.5, None, 4.0, 5.5, 6.0, None, 8.0, 9.5, 10.0],
    'num2': [10.0, None, 30.0, 40.0, 50.0, None, 70.0, 80.0, 90.0, 100.0],
    'category1': ['A', 'B', 'A', None, 'C', 'B', 'C', 'A', 'B', None],
    'category2': ['C', 'A', 'B', None, 'C', 'C', 'B', 'B', 'A', None]
})
df
```

```python
df.info()
```

```python
df['category'] = df['category'].astype('str') #Convert category column to string type
```

```python
# imputer =  SimpleImputer(strategy='mean')
# imputer =  SimpleImputer(strategy='constant', fill_value=0)
imputer =  SimpleImputer(strategy='constant',fill_value='unknown')
imputer.fit_transform(df) #it accepts 2D
# if object dt it checks for NAN (not a number) and replaces it with 'unknown' in this case.
# imputer =  SimpleImputer(strategy='most_frequent') for this it assigns most frequent value
```
